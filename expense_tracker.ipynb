A simple command-line expense tracker that lets a user log expenses, view them by category,
see a running total, and delete entries. Built to apply the Python fundamentals from Module 1:
functions, conditionals, loops, and lists/dictionaries, plus basic file handling to persist data
between sessions using CSV.

## Features
- Add an expense (amount, category, description)
- View all expenses
- View total spending, and totals grouped by category
- Delete an expense by its ID
- Save expenses to `expenses.csv` and load them back on startup


## 1. Imports and data storage setup

```python
import csv
import os
from datetime import date

DATA_FILE = "expenses.csv"
FIELDNAMES = ["id", "date", "amount", "category", "description"]

expenses = []  # each entry: {"id": int, "date": str, "amount": float, "category": str, "description": str}
```

## 2. Load and save helpers

```python
def load_expenses():
    """Load expenses from the CSV file into the expenses list, if the file exists."""
    if not os.path.exists(DATA_FILE):
        return []
    loaded = []
    with open(DATA_FILE, newline="") as f:
        reader = csv.DictReader(f)
        for row in reader:
            row["id"] = int(row["id"])
            row["amount"] = float(row["amount"])
            loaded.append(row)
    return loaded


def save_expenses():
    """Write the current expenses list to the CSV file."""
    with open(DATA_FILE, "w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=FIELDNAMES)
        writer.writeheader()
        writer.writerows(expenses)
```

## 3. Core functions

```python
def next_id():
    return max((e["id"] for e in expenses), default=0) + 1


def add_expense(amount, category, description):
    entry = {
        "id": next_id(),
        "date": date.today().isoformat(),
        "amount": round(float(amount), 2),
        "category": category.strip().title(),
        "description": description.strip(),
    }
    expenses.append(entry)
    save_expenses()
    print(f"Added expense #{entry['id']}: Rs {entry['amount']} - {entry['category']}")


def view_expenses():
    if not expenses:
        print("No expenses recorded yet.")
        return
    print(f"{'ID':<4}{'Date':<12}{'Amount':<10}{'Category':<15}{'Description'}")
    print("-" * 60)
    for e in expenses:
        print(f"{e['id']:<4}{e['date']:<12}{e['amount']:<10}{e['category']:<15}{e['description']}")


def total_spending():
    return round(sum(e["amount"] for e in expenses), 2)


def totals_by_category():
    totals = {}
    for e in expenses:
        totals[e["category"]] = totals.get(e["category"], 0) + e["amount"]
    return {k: round(v, 2) for k, v in totals.items()}


def delete_expense(expense_id):
    global expenses
    before = len(expenses)
    expenses = [e for e in expenses if e["id"] != int(expense_id)]
    if len(expenses) < before:
        save_expenses()
        print(f"Deleted expense #{expense_id}.")
    else:
        print(f"No expense found with ID {expense_id}.")
```

## 4. Demo run
Since this project is normally interactive (via `input()`), this cell demonstrates the core functions directly so the notebook runs end to end without manual typing.

```python
expenses = load_expenses()

add_expense(1500, "Food", "Groceries for the week")
add_expense(800, "Transport", "Bus and rickshaw fares")
add_expense(3000, "Education", "Online course subscription")
add_expense(450, "Food", "Lunch with friends")

print()
view_expenses()

print()
print("Total spending: Rs", total_spending())

print()
print("Totals by category:")
for category, amount in totals_by_category().items():
    print(f"  {category}: Rs {amount}")

print()
delete_expense(2)
view_expenses()
```
